```python
import pandas as pd
import numpy as np

np.random.seed(42)
calisan_sayisi = 200
departmanlar = ['Software', 'HR', 'Finance', 'Sales', 'Marketing'] 

data = {
    'EmployeeId': range(1, calisan_sayisi + 1),
    'Department': np.random.choice(departmanlar, calisan_sayisi),
    'Age': np.random.randint(22, 60, calisan_sayisi),
    'ExperienceYears': np.random.randint(0, 20, calisan_sayisi),
    'EducationLevel': np.random.choice(['Lisans', 'Yüksek Lisans', 'Doktora'], calisan_sayisi),
    'Salary': np.random.randint(40000, 150000, calisan_sayisi),
    'PerformanceScore': np.random.randint(1, 101, calisan_sayisi), 
    'OvertimeHours': np.random.randint(0, 50, calisan_sayisi),
    'ProjectsCompleted': np.random.randint(0, 30, calisan_sayisi)
}

df = pd.DataFrame(data)

toplam_calisan = len(df)
toplam_departman = df['Department'].nunique()
ortalama_maas = df['Salary'].mean()
ortalama_deneyim = df['ExperienceYears'].mean()
ortalama_performans = df['PerformanceScore'].mean()

print(f"Kaç çalışan var?: {toplam_calisan}")
print(f"Kaç departman var?: {toplam_departman}")
print(f"Ortalama maaş ne?: {ortalama_maas:.2f} TL")
print(f"Ortalama deneyim kaç yıl?: {ortalama_deneyim:.1f} Yıl")
print(f"Ortalama performans puanı kaç?: {ortalama_performans:.1f}")
```

```python
departman_analizi = df.groupby('Department').agg(
    CalisanSayisi=('EmployeeId', 'count'),
    OrtalamaMaas=('Salary', 'mean'),
    OrtalamaPerformans=('PerformanceScore', 'mean'),
    OrtalamaDeneyim=('ExperienceYears', 'mean')
).round(1) 

print("Departman Bazlı Analiz Raporu:\n")
print(departman_analizi)
```

```python
def seviye_belirle(yil):
    if yil <= 2:
        return "Junior"
    elif yil <= 5:
        return "Mid"
    elif yil <= 10:
        return "Senior"
    else:
        return "Expert"

df['ExperienceLevel'] = df['ExperienceYears'].apply(seviye_belirle)
df['SalaryPerExperience'] = df['Salary'] / df['ExperienceYears'].replace(0, 1)
print(df[['ExperienceYears', 'ExperienceLevel', 'Salary', 'SalaryPerExperience']].head(10))
```

```python
deneyim_analizi = df.groupby('ExperienceLevel')[['Salary', 'PerformanceScore']].mean()
deneyim_analizi = deneyim_analizi.reindex(['Junior', 'Mid', 'Senior', 'Expert']).round(1)

print("Deneyim Seviyesine Göre Maaş ve Performans:")
print(deneyim_analizi, "\n")

df['MesaiDurumu'] = df['OvertimeHours'].apply(lambda x: 'Fazla Mesai' if x > 20 else 'Az/Yok')
mesai_performans = df.groupby('MesaiDurumu')['PerformanceScore'].mean().round(1)

print("Mesai Yoğunluğuna Göre Ortalama Performans:")
print(mesai_performans, "\n")

df['ProjeDurumu'] = df['ProjectsCompleted'].apply(lambda x: 'Çok Proje' if x > 15 else 'Az Proje')
proje_maas = df.groupby('ProjeDurumu')['Salary'].mean().round(1)

print("Proje Sayısına Göre Ortalama Maaş:")
print(proje_maas)
```

```python
seviye_haritasi = {'Junior': 0, 'Mid': 1, 'Senior': 2, 'Expert': 3}
df['ExperienceLevel'] = df['ExperienceLevel'].map(seviye_haritasi)
df = pd.get_dummies(df, columns=['Department'], dtype=int)

print("Encoding Sonrası Veri Seti (İlk 5 Satır):")
print(df[['EmployeeId', 'ExperienceLevel', 'Department_Software', 'Department_HR', 'Department_Finance']].head())
```

```python
from sklearn.utils import resample
dengesiz_veri = pd.DataFrame({
    'Department': ['Software']*70 + ['HR']*5 + ['Finance']*5,
    'EmployeeId': range(1, 81)
})

print("Resampling Öncesi Dağılım:")
print(dengesiz_veri['Department'].value_counts())
print("-" * 30)

df_software = dengesiz_veri[dengesiz_veri['Department'] == 'Software']
df_hr = dengesiz_veri[dengesiz_veri['Department'] == 'HR']
df_finance = dengesiz_veri[dengesiz_veri['Department'] == 'Finance']

df_hr_cogaltilmis = resample(df_hr, 
                             replace=True,     
                             n_samples=70,    
                             random_state=42)

df_finance_cogaltilmis = resample(df_finance, 
                                  replace=True,     
                                  n_samples=70,     
                                  random_state=42)


dengeli_veri = pd.concat([df_software, df_hr_cogaltilmis, df_finance_cogaltilmis])

print("Resampling Sonrası Dağılım:")
print(dengeli_veri['Department'].value_counts())
```
